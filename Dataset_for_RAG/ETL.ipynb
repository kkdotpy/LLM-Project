```python
import json
import pandas as pd

with open(r'C:\Users\khare\Desktop\LLM_FINAL_PROJ\LLM-Project\Dataset_for_RAG\foodkeeper_USDP.json', 'r') as f:
    data = json.load(f)

# Extract sheets
categories_data = data['sheets'][1]['data']  # Sheet 1
products_data = data['sheets'][2]['data']    # Sheet 2  
safety_data = data['sheets'][3]['data']      # Sheet 3
cooking_data = data['sheets'][4]['data']     # Sheet 4
```

```python
import json
import re
import pandas as pd

def flatten_sheet(sheet_data):
    """Convert nested list-of-dicts to flat dict and return as DataFrame."""
    flattened = []
    for record in sheet_data:
        flat = {}
        for field_dict in record:
            for key, value in field_dict.items():
                flat[key] = value
        flattened.append(flat)
    return pd.DataFrame(flattened)

# Load and flatten all sheets
## Changing the structure from [[{'ID': 1}, {'Name': 'Butter'}, {'Keywords': 'Butter'}], ...] --> [{'ID': 1, 'Name': 'Butter', 'Keywords': 'Butter'}, ...] --> DataFrame
categories_df = flatten_sheet(data['sheets'][1]['data'])
products_df = flatten_sheet(data['sheets'][2]['data'])
safety_df = flatten_sheet(data['sheets'][3]['data'])
cooking_df = flatten_sheet(data['sheets'][4]['data'])
```

```python
# Rename columns for clarity
## Because there are multiple ID's in each sheets
## Renaming them to be more specific for merging later

products_df = products_df.rename(columns={'ID': 'Product_ID'})
safety_df = safety_df.rename(columns={'ID': 'Safety_ID', 'Product_ID': 'Product_ID'})
cooking_df = cooking_df.rename(columns={'ID': 'Cooking_ID', 'Product_ID': 'Product_ID'})
```

```python
# Add category names to products
## Categories has ID and Name, Products has Product Id and category ID, merging them to get category name in products dataframe
products_with_cat = products_df.merge(
    categories_df, 
    left_on='Category_ID', 
    right_on='ID', 
    how='left'
)

products_with_cat.iloc[15:45]
```

```python
def get_specific_name(food_name, keywords):
    """Extract specific variety from keywords
       Essential for varieties of meats cuts, to prevent duplication and for more clarity in the documents. For example, if the food item is 'Beef' and the keywords contain 'rib roast',
     we can name the document 'Beef (Rib Roast)' instead of just 'Beef', which gives more specific information about the product.
    """
    if pd.isna(keywords) or keywords == food_name:
        return food_name
    keywords_lower = str(keywords).lower()
    
    # take first unique keyword
    kw_list = [k.strip() for k in keywords_lower.split(',')]
    for kw in kw_list:
        if kw != food_name.lower() and len(kw) > 2:
            return f"{food_name} ({kw.title()})"
    
    return food_name
```

```python
# === STORAGE DOCUMENTS (661 records) ===
storage_docs = []

for _, row in products_with_cat.iterrows():
    food_name = row.get('Name', 'Unknown') ## Getting the name of the food item, if not present, default to 'Unknown'
    if pd.isna(food_name):
        continue

    keywords = row.get('Keywords', '')
    specific_name = get_specific_name(food_name, keywords)  ## Getting the specific name of the food item based on its keywords, for more clarity in the documents. For example, if the food item is 'Beef' and the keywords contain 'rib roast', we can name the document 'Beef (Rib Roast)' instead of just 'Beef', which gives more specific information about the product.
    
    # Build rich storage text
    ## Idea is to get Pantry info, Refrigerator info (from purchase and after opening) and combine them into a single text field for each product

    storage_parts = []
    
    # Pantry info  ## Some products have pantry tips, if present, we add it to the storage info.
    if pd.notna(row.get('Pantry_tips')):
        storage_parts.append(f"Pantry: {row['Pantry_tips']}")
    
    # Date of Purchase - Refrigerator 
    ## DOP refrigerate min and max gives the range of days for which the product can be stored in the refrigerator from the date of purchase, if both values are present, we add it to the storage info.
    if pd.notna(row.get('DOP_Refrigerate_Min')) and pd.notna(row.get('DOP_Refrigerate_Max')):
        min_val = int(row['DOP_Refrigerate_Min']) if row['DOP_Refrigerate_Min'] % 1 == 0 else row['DOP_Refrigerate_Min']  ## %1 ==0 is just to check if the value is a whole number, if yes, we convert it to int to remove the decimal point, otherwise we keep it as is (could be a fraction or a decimal)
        max_val = int(row['DOP_Refrigerate_Max']) if row['DOP_Refrigerate_Max'] % 1 == 0 else row['DOP_Refrigerate_Max']
        metric = row.get('DOP_Refrigerate_Metric', 'days')
        storage_parts.append(f"Refrigerator (from purchase): {min_val}-{max_val} {metric}")
    
    # After opening
    if pd.notna(row.get('Refrigerate_After_Opening_Min')) and pd.notna(row.get('Refrigerate_After_Opening_Max')):
        min_val = int(row['Refrigerate_After_Opening_Min']) if row['Refrigerate_After_Opening_Min'] % 1 == 0 else row['Refrigerate_After_Opening_Min']
        max_val = int(row['Refrigerate_After_Opening_Max']) if row['Refrigerate_After_Opening_Max'] % 1 == 0 else row['Refrigerate_After_Opening_Max']
        metric = row.get('Refrigerate_After_Opening_Metric', 'days')  ## Getting the metric for after opening, if not present, default to 'days'
        storage_parts.append(f"Refrigerator (after opening): {min_val}-{max_val} {metric}")
    
    if storage_parts:
        search_text = f"{food_name} {row.get('Keywords', '')} {row.get('Category_Name', '')}"  ## Makes it easier to search for the product based on its name, keywords and category. Essentially added for RAG retrieval purposes, so that the relevant document can be retrieved based on the query.
        
        storage_docs.append({
            'id': f"storage_{row['Product_ID']}",
            'food_name': specific_name,
            'category': row.get('Category_Name'),
            'subcategory': row.get('Subcategory_Name'),
            'content': f"{keywords}. " + " | ".join(storage_parts),  ## Joining the different storage parts with a seperator. Include important contents in one line
            'search_text': search_text.lower(),
            'type': 'storage'
        })
```

```python
# === SAFETY DOCUMENTS ===
safety_docs = []

# Merge safety with products to get food names
safety_with_names = safety_df.merge(
    products_df[['Product_ID', 'Name','Keywords' ,'Category_ID']], 
    on='Product_ID', 
    how='left'
)

for _, row in safety_with_names.iterrows():
    food_name = row.get('Name', 'Unknown')
    if pd.isna(food_name):
        continue
    
    keywords = row.get('Keywords', '')
    specific_name = get_specific_name(food_name, keywords)  ## Getting the specific name of
    safety_parts = []
    
    if pd.notna(row.get('Tips')):
        safety_parts.append(f"Tip: {row['Tips']}")
    
    if pd.notna(row.get('Safe_Minimum_Temperature')):
        temp = int(row['Safe_Minimum_Temperature']) if row['Safe_Minimum_Temperature'] % 1 == 0 else row['Safe_Minimum_Temperature']
        safety_parts.append(f"Cook to {temp}°F")
        
        if pd.notna(row.get('Rest_Time')):
            rest = int(row['Rest_Time']) if row['Rest_Time'] % 1 == 0 else row['Rest_Time']
            metric = row.get('Rest_Time_metric', 'minutes')  ## Getting the metric for rest time, if not present, default to 'minutes'
            safety_parts.append(f"Rest for {rest} {metric}")
    
    if safety_parts:
        safety_docs.append({
            'id': f"safety_{row['Product_ID']}",
            'food_name': specific_name,
            'content': f"{specific_name}. " + " | ".join(safety_parts),
            'search_text': f"{specific_name} temperature cook safe".lower(),
            'type': 'safety'
        })
```

```python
# === COOKING DOCUMENTS ===
cooking_docs = []

cooking_with_names = cooking_df.merge(
    products_df[['Product_ID', 'Name', 'Keywords', 'Category_ID']], 
    on='Product_ID', 
    how='left'
)

for _, row in cooking_with_names.iterrows():
    food_name = row.get('Name', 'Unknown')
    if pd.isna(food_name):
        continue

    keywords = row.get('Keywords', '')
    specific_name = get_specific_name(food_name, keywords)
    cooking_parts = []
    
    if pd.notna(row.get('Cooking_Method')):
        cooking_parts.append(f"Method: {row['Cooking_Method']}")
    
    # Temperature
    if pd.notna(row.get('Cooking_Temperature')):
        temp_value = str(row['Cooking_Temperature'])
        cooking_parts.append(f"Temperature: {temp_value}°F")
    
    # Timing (handles from/to)
    if pd.notna(row.get('Timing_from')) and pd.notna(row.get('Timing_to')):
        time_from = row['Timing_from']
        time_to = row['Timing_to']
        metric = row.get('Timing_metric', 'minutes')
        
        # Clean up decimal .0 if present
        if isinstance(time_from, float) and time_from.is_integer():
            time_from = int(time_from)
        if isinstance(time_to, float) and time_to.is_integer():
            time_to = int(time_to)
        
        if pd.notna(row.get('Timing_per')):
            cooking_parts.append(f"Cook {time_from}-{time_to} {metric} per {row['Timing_per']}")
        else:
            cooking_parts.append(f"Cook for {time_from}-{time_to} {metric}")
    
    # Size/Measure (handles ranges)
    if pd.notna(row.get('Measure_from')) and pd.notna(row.get('Measure_to')):
        size_from = row['Measure_from']
        size_to = row['Measure_to']
        metric = row.get('Size_metric', 'pounds')
        
        if isinstance(size_from, float) and size_from.is_integer():
            size_from = int(size_from)
        if isinstance(size_to, float) and size_to.is_integer():
            size_to = int(size_to)
            
        cooking_parts.append(f"For {size_from}-{size_to} {metric} size")
    
    if cooking_parts:
        cooking_docs.append({
            'id': f"cooking_{row['Product_ID']}",
            'food_name': specific_name,
            'content': f"{keywords}. " + " | ".join(cooking_parts),
            'type': 'cooking'
        })

print(f"Cooking docs created: {len(cooking_docs)}")

```

```python
# Saving for RAG
def save_collection(docs, filename):
    with open(filename, 'w', encoding='utf-8') as f:
        for doc in docs:
            f.write(json.dumps(doc, ensure_ascii=False) + '\n')
    print(f"Saved {len(docs)} documents to {filename}")

save_collection(storage_docs, 'rag_storage.jsonl')
save_collection(safety_docs, 'rag_safety.jsonl')
save_collection(cooking_docs, 'rag_cooking.jsonl')

print("\n=== RAG Collections Created ===")
print(f"Storage docs: {len(storage_docs)}")
print(f"Safety docs: {len(safety_docs)}")
print(f"Cooking docs: {len(cooking_docs)}")
```
